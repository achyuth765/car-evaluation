#### Car Evaluation Classification

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json

from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                              confusion_matrix, classification_report)

sns.set_style("whitegrid")
plt.rcParams['figure.dpi'] = 110
pd.set_option("display.precision", 4)

RANDOM_STATE = 42
```

##### 1. Data Exploration and Pre-processing

```python
cols = ["buying", "maint", "doors", "persons", "lug_boot", "safety", "class"]
df = pd.read_csv("car.data", names=cols)

print(f"Number of observations: {df.shape[0]}")
print(f"Number of input features: {df.shape[1] - 1}")
print()
print("Data types:")
print(df.dtypes)
```

```python
df.head(10)
```

```python

orig_counts = df["class"].value_counts()
orig_pct = df["class"].value_counts(normalize=True).mul(100).round(2)
pd.DataFrame({"Count": orig_counts, "Percentage (%)": orig_pct})
```

```python
mapping = {"unacc": "Unacceptable", "acc": "Acceptable", "good": "Good", "vgood": "Good"}
df["target"] = df["class"].map(mapping)

new_counts = df["target"].value_counts().reindex(["Unacceptable","Acceptable","Good"])
new_pct = df["target"].value_counts(normalize=True).reindex(["Unacceptable","Acceptable","Good"]).mul(100).round(2)
pd.DataFrame({"Count": new_counts, "Percentage (%)": new_pct})
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))

orig_order = orig_counts.reindex(["unacc","acc","good","vgood"])
axes[0].bar(orig_order.index, orig_order.values, color=["#4C72B0","#55A868","#C44E52","#8172B2"])
axes[0].set_title("Original 4-Class Distribution")
axes[0].set_ylabel("Count")
for i, v in enumerate(orig_order.values):
    axes[0].text(i, v + 15, str(v), ha="center", fontsize=9)

axes[1].bar(new_counts.index, new_counts.values, color=["#4C72B0","#55A868","#C44E52"])
axes[1].set_title("Merged 3-Class Distribution (good + vgood -> Good)")
axes[1].set_ylabel("Count")
for i, v in enumerate(new_counts.values):
    axes[1].text(i, v + 15, str(v), ha="center", fontsize=9)

plt.tight_layout()
plt.show()
```

**Observation:** the dataset is imbalanced — `Unacceptable` accounts for 70% of
observations while the merged `Good` class is under 8%. A model that always predicts
`Unacceptable` would already score 70% accuracy, so macro-averaged metrics and the
confusion matrix are used throughout (not just overall accuracy) to make sure the
minority classes are properly evaluated.

```python
buying_order   = {"low": 0, "med": 1, "high": 2, "vhigh": 3}
maint_order    = {"low": 0, "med": 1, "high": 2, "vhigh": 3}
doors_order    = {"2": 0, "3": 1, "4": 2, "5more": 3}
persons_order  = {"2": 0, "4": 1, "more": 2}
lug_boot_order = {"small": 0, "med": 1, "big": 2}
safety_order   = {"low": 0, "med": 1, "high": 2}
target_order   = {"Unacceptable": 0, "Acceptable": 1, "Good": 2}

df_enc = df.copy()
df_enc["buying"]     = df_enc["buying"].map(buying_order)
df_enc["maint"]      = df_enc["maint"].map(maint_order)
df_enc["doors"]      = df_enc["doors"].map(doors_order)
df_enc["persons"]    = df_enc["persons"].map(persons_order)
df_enc["lug_boot"]   = df_enc["lug_boot"].map(lug_boot_order)
df_enc["safety"]     = df_enc["safety"].map(safety_order)
df_enc["target_enc"] = df_enc["target"].map(target_order)

feature_cols = ["buying", "maint", "doors", "persons", "lug_boot", "safety"]
class_names = ["Unacceptable", "Acceptable", "Good"]

df_enc[feature_cols + ["target_enc"]].head()
```

```python
X = df_enc[feature_cols]
y = df_enc["target_enc"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)

print("Train shape:", X_train.shape, " Test shape:", X_test.shape)
print()
print("Train class proportions:\n", y_train.value_counts(normalize=True).sort_index())
print()
print("Test class proportions:\n", y_test.value_counts(normalize=True).sort_index())
```

```python
fig, ax = plt.subplots(figsize=(7, 4.5))
train_pct = y_train.value_counts(normalize=True).reindex([0,1,2]) * 100
test_pct  = y_test.value_counts(normalize=True).reindex([0,1,2]) * 100
x = np.arange(3); width = 0.35

ax.bar(x - width/2, train_pct.values, width, label="Train (80%)", color="#4C72B0")
ax.bar(x + width/2, test_pct.values, width, label="Test (20%)", color="#DD8452")
ax.set_xticks(x); ax.set_xticklabels(class_names)
ax.set_ylabel("Percentage (%)")
ax.set_title("Class Distribution: Train vs Test (Stratified Split)")
ax.legend()
for i, (tr, te) in enumerate(zip(train_pct.values, test_pct.values)):
    ax.text(i - width/2, tr + 1, f"{tr:.1f}%", ha="center", fontsize=8)
    ax.text(i + width/2, te + 1, f"{te:.1f}%", ha="center", fontsize=8)
plt.tight_layout()
plt.show()
```

```python
orders = {
    "buying": ["low","med","high","vhigh"], "maint": ["low","med","high","vhigh"],
    "doors": ["2","3","4","5more"], "persons": ["2","4","more"],
    "lug_boot": ["small","med","big"], "safety": ["low","med","high"],
}
target_colors = {"Unacceptable": "#4C72B0", "Acceptable": "#55A868", "Good": "#C44E52"}

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
axes = axes.flatten()
for ax, feat in zip(axes, feature_cols):
    ct = pd.crosstab(df[feat], df["target"], normalize="index").reindex(orders[feat])
    ct = ct[["Unacceptable","Acceptable","Good"]]
    ct.plot(kind="bar", stacked=True, ax=ax, color=[target_colors[c] for c in ct.columns], legend=False)
    ax.set_title(f"Target proportion by {feat}")
    ax.set_xlabel(feat); ax.set_ylabel("Proportion")
    ax.tick_params(axis="x", rotation=0)
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="upper center", ncol=3, bbox_to_anchor=(0.5, 1.03))
plt.tight_layout()
plt.show()
```

**Key patterns:**
- `safety = low` always results in `Unacceptable` — no exceptions.
- `persons = 2` (a 2-seater) is also always `Unacceptable`, regardless of other attributes.
- Higher `buying`/`maint` prices shift the distribution toward `Unacceptable`.
- `doors` shows the weakest visible relationship with the target.


```python
fig, ax = plt.subplots(figsize=(6.5, 5.5))
corr = df_enc[feature_cols + ["target_enc"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0, fmt=".2f", ax=ax)
ax.set_title("Correlation Heatmap (Ordinal-Encoded Features & Target)")
plt.tight_layout()
plt.show()
```

#### 2. Baseline Decision Tree Model

```python
clf = DecisionTreeClassifier(criterion="gini", random_state=RANDOM_STATE)
clf.fit(X_train, y_train)

y_pred = clf.predict(X_test)

acc          = accuracy_score(y_test, y_pred)
prec_macro   = precision_score(y_test, y_pred, average="macro")
rec_macro    = recall_score(y_test, y_pred, average="macro")
f1_macro     = f1_score(y_test, y_pred, average="macro")
prec_weight  = precision_score(y_test, y_pred, average="weighted")
rec_weight   = recall_score(y_test, y_pred, average="weighted")
f1_weight    = f1_score(y_test, y_pred, average="weighted")

print(f"Accuracy:              {acc:.4f}")
print(f"Precision (macro):     {prec_macro:.4f}")
print(f"Recall (macro):        {rec_macro:.4f}")
print(f"F1-score (macro):      {f1_macro:.4f}")
print(f"Precision (weighted):  {prec_weight:.4f}")
print(f"Recall (weighted):     {rec_weight:.4f}")
print(f"F1-score (weighted):   {f1_weight:.4f}")
print(f"Tree depth:            {clf.get_depth()}")
print(f"Number of leaves:      {clf.get_n_leaves()}")
```

```python
print(classification_report(y_test, y_pred, target_names=class_names, digits=3))
```

```python
cm = confusion_matrix(y_test, y_pred)
fig, ax = plt.subplots(figsize=(5.5, 4.8))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=class_names, yticklabels=class_names, ax=ax)
ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
ax.set_title("Baseline Decision Tree — Confusion Matrix")
plt.tight_layout()
plt.show()
```

**Interpretation:** the full tree reaches ~98% accuracy, far above the 70%
majority-class baseline. `Unacceptable` and `Good` are almost perfectly separated,
but `Acceptable` is the hardest class — it sits in the middle of the ordinal scale,
bordering both other classes, so attribute combinations near the boundary are more
easily confused.

### 4. Tree Complexity and Bias–Variance Trade-off
Decision Trees are retrained with max_depth from 1 to 15, recording train/test accuracy and macro-F1 at each depth.

```python
depths = list(range(1, 16))
train_acc, test_acc, train_f1, test_f1 = [], [], [], []

for d in depths:
    dtree = DecisionTreeClassifier(criterion="gini", max_depth=d, random_state=RANDOM_STATE)
    dtree.fit(X_train, y_train)
    tr_pred, te_pred = dtree.predict(X_train), dtree.predict(X_test)
    train_acc.append(accuracy_score(y_train, tr_pred))
    test_acc.append(accuracy_score(y_test, te_pred))
    train_f1.append(f1_score(y_train, tr_pred, average="macro"))
    test_f1.append(f1_score(y_test, te_pred, average="macro"))

depth_results = pd.DataFrame({
    "depth": depths, "train_acc": train_acc, "test_acc": test_acc,
    "train_f1_macro": train_f1, "test_f1_macro": test_f1
})
best_depth_idx = int(np.argmax(test_acc))
best_depth = depths[best_depth_idx]
print(f"Best depth by test accuracy: {best_depth}  (test acc = {test_acc[best_depth_idx]:.4f})")
depth_results
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))
axes[0].plot(depths, train_acc, marker="o", label="Train Accuracy", color="#4C72B0")
axes[0].plot(depths, test_acc, marker="s", label="Test Accuracy", color="#DD8452")
axes[0].axvline(best_depth, color="green", linestyle="--", alpha=0.6, label=f"Best depth={best_depth}")
axes[0].set_xlabel("Max Tree Depth"); axes[0].set_ylabel("Accuracy")
axes[0].set_title("Accuracy vs Tree Depth"); axes[0].legend(); axes[0].set_xticks(depths)

axes[1].plot(depths, train_f1, marker="o", label="Train Macro-F1", color="#4C72B0")
axes[1].plot(depths, test_f1, marker="s", label="Test Macro-F1", color="#DD8452")
axes[1].axvline(best_depth, color="green", linestyle="--", alpha=0.6, label=f"Best depth={best_depth}")
axes[1].set_xlabel("Max Tree Depth"); axes[1].set_ylabel("Macro F1-score")
axes[1].set_title("Macro F1-score vs Tree Depth"); axes[1].legend(); axes[1].set_xticks(depths)
plt.tight_layout()
plt.show()
```

**Bias–variance trade-off:**
Underfitting (depth 1–3): train and test accuracy are both low and close
  together — the model is too simple (high bias) to separate `Acceptable` from `Good`.
  Good fit (depth ≈ 6–8): train and test performance rise together and stay
  close; test accuracy peaks around **depth 8**.
  Overfitting (depth ≥ 10): training accuracy reaches 100% by depth 12, but test
  accuracy plateaus (and dips very slightly below the depth-8 peak) — the extra depth
  fits noise/idiosyncrasies in the training rows (high variance) rather than
  generalisable rules.

The best generalisation is achieved at a **moderate depth (8)**, which should be
preferred over the unconstrained full tree: it matches or slightly exceeds its test
performance with far less complexity.


```python
best_dt = DecisionTreeClassifier(criterion="gini", max_depth=best_depth, random_state=RANDOM_STATE)
best_dt.fit(X_train, y_train)
y_pred_best_dt = best_dt.predict(X_test)

best_dt_metrics = {
    "depth": best_depth,
    "test_accuracy": accuracy_score(y_test, y_pred_best_dt),
    "test_precision_macro": precision_score(y_test, y_pred_best_dt, average="macro"),
    "test_recall_macro": recall_score(y_test, y_pred_best_dt, average="macro"),
    "test_f1_macro": f1_score(y_test, y_pred_best_dt, average="macro"),
}
best_dt_metrics
```

## 5. Random Forest: Bagging Ensemble Learning

```python
param_grid_rf = {
    "n_estimators": [100, 200, 300],
    "max_depth": [8, 10, 12, None],
    "max_features": ["sqrt", "log2", None],
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
rf_base = RandomForestClassifier(random_state=RANDOM_STATE, oob_score=True, bootstrap=True, n_jobs=-1)

grid_rf = GridSearchCV(rf_base, param_grid_rf, cv=cv, scoring="f1_macro", n_jobs=-1)
grid_rf.fit(X_train, y_train)

print("Best params:", grid_rf.best_params_)
print(f"Best CV macro-F1: {grid_rf.best_score_:.4f}")

best_rf = grid_rf.best_estimator_
print(f"OOB accuracy: {best_rf.oob_score_:.4f}")
```

```python
y_pred_rf = best_rf.predict(X_test)

rf_metrics = {
    "test_accuracy": accuracy_score(y_test, y_pred_rf),
    "test_precision_macro": precision_score(y_test, y_pred_rf, average="macro"),
    "test_recall_macro": recall_score(y_test, y_pred_rf, average="macro"),
    "test_f1_macro": f1_score(y_test, y_pred_rf, average="macro"),
}
print(json.dumps(rf_metrics, indent=2))
print()
print(classification_report(y_test, y_pred_rf, target_names=class_names, digits=3))
```

```python
cm_rf = confusion_matrix(y_test, y_pred_rf)
fig, ax = plt.subplots(figsize=(5.5, 4.8))
sns.heatmap(cm_rf, annot=True, fmt="d", cmap="Greens", xticklabels=class_names, yticklabels=class_names, ax=ax)
ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
ax.set_title("Tuned Random Forest — Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
best_max_depth = grid_rf.best_params_["max_depth"]
best_max_features = grid_rf.best_params_["max_features"]
n_est_range = [10, 20, 30, 50, 75, 100, 150, 200, 300, 400, 500]
oob_errors = []

for n in n_est_range:
    rf_n = RandomForestClassifier(n_estimators=n, max_depth=best_max_depth,
                                   max_features=best_max_features, oob_score=True,
                                   bootstrap=True, random_state=RANDOM_STATE, n_jobs=-1)
    rf_n.fit(X_train, y_train)
    oob_errors.append(1 - rf_n.oob_score_)

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(n_est_range, oob_errors, marker="o", color="#C44E52")
ax.set_xlabel("Number of Trees (n_estimators)")
ax.set_ylabel("OOB Error Rate")
ax.set_title(f"OOB Error vs Number of Trees (max_depth={best_max_depth}, max_features={best_max_features})")
plt.tight_layout()
plt.show()
```

Random Forest's main practical benefit here is robustness — less sensitivity to hyperparameter choice, reduced overfitting risk if depth were left unconstrained, and a free OOB error estimate — rather than a large accuracy gain.

### 6. Boosting: Building Strong Learners from Weak Learners 

```python
base_stump = DecisionTreeClassifier(max_depth=1, random_state=RANDOM_STATE)

param_grid_ada = {
    "n_estimators": [50, 100, 150, 200, 300],
    "learning_rate": [0.01, 0.05, 0.1, 0.5, 1.0],
}

ada_base = AdaBoostClassifier(estimator=base_stump, random_state=RANDOM_STATE)
grid_ada = GridSearchCV(ada_base, param_grid_ada, cv=cv, scoring="f1_macro", n_jobs=-1)
grid_ada.fit(X_train, y_train)

print("Best params:", grid_ada.best_params_)
print(f"Best CV macro-F1: {grid_ada.best_score_:.4f}")

best_ada = grid_ada.best_estimator_
```

```python
y_pred_ada = best_ada.predict(X_test)

ada_metrics = {
    "test_accuracy": accuracy_score(y_test, y_pred_ada),
    "test_precision_macro": precision_score(y_test, y_pred_ada, average="macro"),
    "test_recall_macro": recall_score(y_test, y_pred_ada, average="macro"),
    "test_f1_macro": f1_score(y_test, y_pred_ada, average="macro"),
}
print(json.dumps(ada_metrics, indent=2))
print()
print(classification_report(y_test, y_pred_ada, target_names=class_names, digits=3))
```

```python
cm_ada = confusion_matrix(y_test, y_pred_ada)
fig, ax = plt.subplots(figsize=(5.5, 4.8))
sns.heatmap(cm_ada, annot=True, fmt="d", cmap="Oranges", xticklabels=class_names, yticklabels=class_names, ax=ax)
ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
ax.set_title("Tuned AdaBoost — Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
staged_train_acc = list(best_ada.staged_score(X_train, y_train))
staged_test_acc  = list(best_ada.staged_score(X_test, y_test))

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(range(1, len(staged_train_acc)+1), staged_train_acc, label="Train Accuracy", color="#4C72B0")
ax.plot(range(1, len(staged_test_acc)+1), staged_test_acc, label="Test Accuracy", color="#DD8452")
ax.set_xlabel("Number of Boosting Rounds (Weak Learners)")
ax.set_ylabel("Accuracy")
ax.set_title("AdaBoost: Accuracy vs Number of Estimators\n(illustrates sequential error correction)")
ax.legend()
plt.tight_layout()
plt.show()
```

AdaBoost clearly underperforms both the Decision Tree and Random Forest, struggling most on the minority Acceptable and Good classes. The likely reason is that a depth-1 stump can only split on one feature at a time, but this dataset's true decision rule is inherently conjunctive (e.g. "safety is high AND persons ≥ 4 AND buying is low AND maint is low" for a Good rating) — reconstructing such interactions from a weighted sum of single-feature stumps is a much harder representational task than a single tree, which encodes multi-feature conjunctions directly along one root-to-leaf path. AdaBoost is also known to be sensitive to class imbalance, since it does not explicitly rebalance classes.

### 6.3 Bagging vs. Boosting for This Problem

**Bagging (Random Forest) is more suitable than boosting (AdaBoost with stumps)**
for this car evaluation problem:
- Random Forest outperforms AdaBoost on every metric measured.
- Random Forest makes far fewer errors on the imbalanced minority classes.
- This is a small, noise-free, deterministic, rule-based dataset with strong feature
  interactions — exactly the setting where a single well-grown tree (and an ensemble
  of such trees) already captures the rules almost perfectly, while boosting's core
  strength (correcting the residual errors of high-bias weak learners) offers little
  extra value when depth-1 stumps are a poor structural match for conjunctive rules.
- Random Forest additionally provides a free OOB error estimate that closely matches
  test performance.

*Caveat:* AdaBoost's weaker performance here is partly a consequence of pairing it
with maximally weak depth-1 stumps, as specified in the brief; deeper weak learners
(e.g. depth-2/3 trees) would likely narrow the gap by letting each learner capture
small feature interactions directly. Under the tuning performed here, though,
bagging is the stronger choice.
